## Problem Statement

Given the lengths

```python
A = 10
B = 7
```

find $\theta_1$ and $\theta_2$ to place the tip of the robot at 
$$^0P_{tip} = \begin{bmatrix} 6.478 \\ 1.142 \\ 12.39\end{bmatrix}$$

- verify your solution with forward kinematics

```python
P_tip_0 = np.array([6.478, 1.142, 12.39])
```

```python
th1 = np.arctan2(P_tip_0[1], P_tip_0[0])*rtd
th1
```

## Use $\theta_1$ to get to a sideview

<img src="https://faculty.gvsu.edu/kraussry/445_images/lecture_slides_images_445/class_13_inv_kin/class_13_inverse_kinematics_part_2_main_markedup_0005.jpg" width=800px>

```python
T01 = DH(0,0,th1, A)
```

```python
T10 = robotics.HTinv(T01)
T10
```

```python
prettymat(T10@T01)
```

```python
P_tip_0 = robotics.append_one(P_tip_0)
```

```python
P_tip_1 = prettymat(T10@P_tip_0)
P_tip_1
```

```python
th2 = np.arctan2(P_tip_1[2],P_tip_1[0])*rtd
th2
```

## $\theta_2$ alternative route

$$^0P_{tipz} = A + B \sin \theta_2$$

$$\sin \theta_2 = (^0P_{tipz} - A)/B$$ 

```python
sin_th2 = (P_tip_0[2]-A)/B
sin_th2
```

```python
cos_th2_p = np.sqrt(1-sin_th2**2)
cos_th2_p
```

```python
th2_alt_pos = np.arctan2(sin_th2, cos_th2_p)*rtd
th2_alt_pos
```
