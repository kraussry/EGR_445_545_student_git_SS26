## What does the DH convention actually do? (Slide 13)

- tells us where to put our frames
    - origin location and unit vector orientation
- tells us how to move from one frame to the next
- models each robot link using four parameters: $\alpha$, $a$, $\theta_i$, and $d$

## DH Principles (Slide 15)

- ${Z}_i$ axes along joint axes
    - rotational joints rotate around ${Z}_i$
    - prismatic joints translate along ${Z}_i$
- ${X}_{i-1}$ axis points along common perpendicular between ${Z}_{i-1}$ and ${Z}_i$ or perpendicular to ${Z}_{i-1}$ and ${Z}_i$ if they intersect
    - you must be able to rotate about ${X}_{i-1}$ to get from ${Z}_{i-1}$ to something parallel to ${Z}_i$
- to get from frame $i-1$ to frame $i$, follow these four steps in order:
    1. rotate by $\alpha_{i-1}$ about ${X}_{i-1}$
    2. translate by $a_{i-1}$ along  ${X}_{i-1}$
    3. rotate by $\theta_i$ about ${Z}_i$
    4. translate by $d_i$ along ${Z}_i$


## DH Big Idea (Slide 16)

The frames must be placed so that for each link it is possible to move from frame $i-1$ to frame $i$ by following the four steps in order:


1. rotate by $\alpha_{i-1}$ about ${X}_{i-1}$
2. translate by $a_{i-1}$ along  ${X}_{i-1}$
3. rotate by $\theta_i$ about ${Z}_i$
4. translate by $d_i$ along ${Z}_i$

- This seems really simple, but $90\%$+ of student mistakes related to DH can be found by checking the four steps in order.


## DH Frame Attachment Summary (Slide 17)

1. Identify the joint axes
2. Identify the common perpendiculars or the point of intersection between the joint axes
3. Assign the ${Z}_i$ axes along the $i^{th}$ joint axes
4. Assign the ${X}_i$ axes along the common perpendiculars or
perpendicular to both ${Z}$ axes if two consecutive ${Z}_i$
intersect
5. Assign the ${Y}$ axes to complete a right-hand coordinate system
    - normally skipped
6. Decide on frames {0} and {N}
