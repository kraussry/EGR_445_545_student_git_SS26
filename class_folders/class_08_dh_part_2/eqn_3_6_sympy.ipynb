# Eqn 3.6 Derivation

```python
import robotics_sympy
```

```python
import sympy
```

```python
alpha, a, th, d = sympy.symbols(['\\alpha', 'a', '\\theta', 'd'])
```

## Claim 1: we can combine the $\hat{X}$ axis HT matrices

$$^{i-1}_Q \mathbf{T}  = \; ^{i-1}_R \mathbf{T} \cdot ^R_Q \mathbf{T} = HTx(\alpha_{i-1}, x=a_{i-1})$$

```python
T_im1R = robotics_sympy.DH(alpha, 0,0,0)
T_im1R
```

```python
T_RQ = robotics_sympy.DH(0,a,0,0)
T_RQ
```

```python
T_im1Q = T_im1R@T_RQ
T_im1Q
```

```python
T_im1Q_combined = robotics_sympy.HTx(alpha, a, 0,0)
T_im1Q_combined
```

## Claim 2: We can similarly combined the $\hat{Z}$ axis steps

```python
T_QP = robotics_sympy.DH(0,0,th,0)
T_QP
```

```python
T_Pi = robotics_sympy.DH(0,0,0,d)
T_Pi
```

```python
T_Qi = T_QP@T_Pi
T_Qi
```

```python
T_Qi_combined = robotics_sympy.HTz(th, 0,0,d)
T_Qi_combined
```

## Eqn. 3.6 comes from multiplying these two together

$$^{i-1}_i\mathbf{T} = \; HT_x(\alpha_{i-1}, x=a_{i-1}) \cdot HTz(\theta_i, z=d_i)$$

```python
T_im1_i = T_im1Q_combined@T_Qi_combined
T_im1_i
```
