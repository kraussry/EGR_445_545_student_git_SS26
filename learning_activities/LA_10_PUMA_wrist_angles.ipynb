```python
import sympy
import robotics_sympy
```

```python
th4s, th5s, th6s, Gs = sympy.symbols(["\\theta_4","\\theta_5", "\\theta_6", "G"])
th4s
```

```python
TW4s = robotics_sympy.DH(0,0,th4s,0)
TW4s
```
